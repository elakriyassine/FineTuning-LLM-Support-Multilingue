```python
import json
import torch
from datasets import Dataset
from peft import LoraConfig, get_peft_model
from rouge_score import rouge_scorer
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    DataCollatorForLanguageModeling,
    Trainer,
    TrainingArguments
)

SEED = 42
torch.manual_seed(SEED)

# Le fichier JSON doit être chargé dans l'espace
# de travail Colab avant de lancer ce script
with open(
    '08_finetune_dataset.json',
    'r',
    encoding='utf-8'
) as file:
    data = json.load(file)


def format_example(example):
    return {
        'text': (
            f"### Entrée:\n{example['input']}\n"
            f"### Rapport:\n{example['output']}"
        )
    }


train_dataset = Dataset.from_list([
    format_example(example)
    for example in data['train']
])
test_dataset = Dataset.from_list([
    format_example(example)
    for example in data['test']
])

model_name = 'TinyLlama/TinyLlama-1.1B-Chat-v1.0'
tokenizer = AutoTokenizer.from_pretrained(model_name)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = 'right'

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map='auto'
)
model.config.use_cache = False

# LoRA adapte seulement les projections utiles
# et garde la majorité des poids d'origine gelés
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=['q_proj', 'v_proj'],
    lora_dropout=0.05,
    bias='none',
    task_type='CAUSAL_LM'
)
model = get_peft_model(model, lora_config)
model.print_trainable_parameters()


def tokenize(batch):
    return tokenizer(
        batch['text'],
        truncation=True,
        max_length=512,
        padding='max_length'
    )


train_tokens = train_dataset.map(
    tokenize,
    batched=True,
    remove_columns=['text']
)
test_tokens = test_dataset.map(
    tokenize,
    batched=True,
    remove_columns=['text']
)
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

training_args = TrainingArguments(
    output_dir='./ids_llm_checkpoints',
    num_train_epochs=3,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    warmup_steps=10,
    logging_steps=10,
    save_strategy='epoch',
    eval_strategy='epoch',
    load_best_model_at_end=True,
    fp16=True,
    seed=SEED,
    data_seed=SEED,
    report_to='none'
)
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_tokens,
    eval_dataset=test_tokens,
    data_collator=data_collator
)
trainer.train()

# Les scores ROUGE comparent les textes générés
# aux rapports de référence du jeu de test
model.config.use_cache = True
scorer = rouge_scorer.RougeScorer(
    ['rouge1', 'rouge2', 'rougeL'],
    use_stemmer=True
)
rouge_scores = {
    'rouge1': [],
    'rouge2': [],
    'rougeL': []
}
generation_device = next(model.parameters()).device

for example in data['test'][:10]:
    prompt = (
        f"### Entrée:\n{example['input']}\n"
        "### Rapport:\n"
    )
    inputs = tokenizer(
        prompt,
        return_tensors='pt',
        max_length=256,
        truncation=True
    )
    inputs = {
        key: value.to(generation_device)
        for key, value in inputs.items()
    }

    torch.manual_seed(SEED)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            temperature=0.7,
            do_sample=True,
            top_p=0.9,
            repetition_penalty=1.2,
            pad_token_id=tokenizer.eos_token_id
        )

    generated = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )
    generated = generated.split(
        '### Rapport:',
        1
    )[-1].strip()
    score = scorer.score(
        example['output'],
        generated
    )
    for metric_name in rouge_scores:
        rouge_scores[metric_name].append(
            score[metric_name].fmeasure
        )

print('\nMétriques ROUGE après fine-tuning :')
for metric_name, values in rouge_scores.items():
    print(
        f'  {metric_name}: '
        f'{sum(values) / len(values):.4f}'
    )

# Le modèle fusionné peut ensuite être chargé
# hors ligne sur CPU sans dépendre de l'adaptateur
model.save_pretrained('./ids_llm_adapter')
merged_model = model.merge_and_unload()
merged_model.save_pretrained(
    './ids_llm_finetuned',
    safe_serialization=True
)
tokenizer.save_pretrained('./ids_llm_finetuned')
print(
    '\nModèle sauvegardé : '
    'télécharger ids_llm_finetuned/'
)
```
